### 3 Major Categories of ML
*  Supervised Learning -> Labeled input and output pairs
*  Unsupervised Learning -> Data without any additional feedback, without labels
*  Reinforcement Learning -> An agent learns from rewards or punishments that it receives

# <b> <font color=#ff4600>1. Supervised Learning</font></b><br>

## <b> <font color=#ff4600>1.1 Classification</font></b><br>


Supervised learning task of learning a function mapping an input point to a <b> <font color='yellow'>discrete category</font> </b><br>
e.g. Given the historical dataset of <b> <font color='green'>[Date],[Humidity],[Pressure]</font> </b> for the computer to predict the weather [Rain] --> Label / Category  <br><br>
<b><font color='green'>f (humidity, pressure)</font></b><br>
f (93, 999) = Rain <br>
f (49, 1010) = No Rain <br>
<b><font color='green'>h (humidity, pressure)</font></b><br>
h = Hypothesis<br><br>
Ideally f and h should be agree with each other as much as possible.

#### <b><font color='lightyellow'> Nearest-Neighber Classification</font></b>
Algorithm that, given an input, choose the class of the nearest data point to that input.<br>

#### <b><font color='lightyellow'> K-Nearest-Neighbor Classification</font></b>
Algorithm that, given an input, choose the most common class out of the k nearest data points to that input. <br>
Weight Vector w: <b><font color='green'>(w₀, w₁, w₂)</font></b> -> goal is to find the optimal weight <br>
Input Vector x: <b><font color='green'>(1, x₁, x₂)</font></b><br>
w.x: <b><font color='green'>w₀+w₁.x₁+w₂.x₂</font></b><br>
h(x₁, x₂) = <b><font color='green'>1 if w₀+w₁.x₁+w₂.x₂ >= 0 </font></b><br>
otherwise h(x₁, x₂) = <b><font color='green'>0</font></b><br>

#### <b><font color='lightyellow'> Perceptron Learning Rule </font></b>
Given data point (x, y), update each weight according to<br>
<b><font color='green'>wᵢ = wᵢ + α(actual value - estimate).xᵢ</font></b><br> 
<b><font color='green'>wᵢ = wᵢ + α(y-hw(x)).xᵢ</font></b><br> 
<b><font color='green'>α = learning rate -> how quick the weight gets updated</font></b><br> 
-> Only capable of learning linearly separable decision boundary

#### <b><font color='lightyellow'> Maximum Margin Separator </font></b>
Boundary that maximizes the distance between any of the data points.

## <b> <font color=#ff4600>1.2 Regression</font></b><br>

Supervised learning task of learning a function mapping an input point to a  <b> <font color='yellow'>continuous value</font> </b><br>
e.g. <br> 
<b><font color='green'>f(advertising)</font></b><br> 
    f(1200) = 5800 (sales)<br>
    f(1800) = 8400 (sales)<br>
<b><font color='green'>h(advertising)</font></b><br> 

#### <b><font color='lightyellow'> Loss Function </font></b>
Function that expresses how poorly our hypothesis performs.

#### <b><font color='lightyellow'> 0-1 Loss Function </font></b><br>
<b><font color='green'>L(actual, predicted):</font></b><br>
<b><font color='green'>0</font></b> if actual = predicted<br>
<b><font color='green'>1</font></b> otherwise<br>

#### <b><font color='lightyellow'> L1 Loss Function </font></b>
<b><font color='green'>L(actual, predicted) = |actual - predicted|</font></b> <br>

#### <b><font color='lightyellow'> L2 Loss Function </font></b>
<b><font color='green'>L(actual, predicted) = (actual - predicted)²</font></b> <br>
<font color='red'>-> It penalises more rashly anything that is worse predicted</font>

#### <b><font color='lightyellow'>Overfitting</font></b>
A model that fits too closely to a particular dataset and therefore may fail to generalize to future data.

#### <b><font color='lightyellow'> Regularization </font></b>
Penalizing hypotheses that are more complex to favor simpler, more general hypotheses<br>
<b><font color='green'>cost(h) = loss(h) + λcomplexity(h)</font></b> <br>


#### <b><font color='lightyellow'> Holdout Cross-Validation </font></b>
Splitting data into a training set and a test set, such that learning happens on the training set and is evaluated on the test set.

#### <b><font color='lightyellow'> K-Fold Cross Validation </font></b>
Splitting data into k sets, and experimenting k times, using each set as a test set once, and using remaining data as training set.

# <b> <font color=#ff4600>2. Unsupervised Learning </font></b><br>

Given input data without any additional feedback, learn patterns

#### <b><font color='lightyellow'> Clustering </font></b>
Organizing a set of objects into groups in such a way that similar objects tend to be in the same group. Some clustering applications: <br>

* Genetic research
* Image segmentation
* Market research
* Medical imaging
* Social network analysis


#### <b><font color='lightyellow'> K-Means Clustering </font></b>
Algorithms for clustering data based on repeatedly assigning points to clusters and updating those clusters' centers.


# <b> <font color=#ff4600>3. Reinforcement Learning </font></b><br>

Given a set of rewards or punishments, learn what actions to take in the future. e.g. game "nim".<br>
Agent can be a physicaly robot or an AI. Reward can be something numerical.<br>



<a href="https://en.wikipedia.org/wiki/Reinforcement_learning">Wikipedia - Reinforcement Learning</a>

#### <b><font color='lightyellow'> Markov Decision Process </font></b>
Model for decision-making, representing states, actions, and their rewards.<br>

* Set of states: S<br>
* Set of actions: ACTIONS (s) <br>
* Transition model: P(s'| s,a) <br>
* Reward Function: R(s,a,s')<br>


<a href="https://inst.eecs.berkeley.edu/~cs188/textbook/mdp/markov-decision-processes.html">Reading: Markov-Decision-Processes - University of California Berkeley </a>

#### <b><font color='lightyellow'> Q-Learning </font></b>
Method for learning function <b><font color='green'>Q(s,a)</font></b>, estimate of the value of performing action <font color='green'>a</font></b> in state <font color='green'>s</font></b>:<br>

* state with Q(s,a) = 0 for all s, a 
* every time we take an action a in state s and observe a reward r, we update:



<a href="https://en.wikipedia.org/wiki/Q-learning">Wikipedia - Q-learning</a>
<br>
* Q(s,a) <- Q(s,a) + α (new value estimiate - old value estimate)
* new value estimate = r (experienced reward) + future reward estimate (take the maximum across all the possible actions I could take next)


#### <b><font color='lightyellow'> Greedy Decision-Making </font></b>
When in state s, choose action a with higher <b><font color='green'>Q(s,a)</font></b>

#### <b><font color='lightyellow'> Explore vs Exploit </font></b>
* Explore: other possibilities
* Exploit: might not be able to maximize the reward (downside of greedy decision-making)

#### <b><font color='lightyellow'> Epsilon-greedy (ε-greedy) </font></b>
* set ε equal to how often we want to move randomly
* with probability <b><font color='green'>1- ε</font></b>, chooese estimated best move
* with probability <b><font color='green'>ε</font></b>, choose a random move



#### <b><font color='lightyellow'> Function Approximation </font></b>
Approximating <b><font color='green'>Q(s,a)</font></b>, often by a function combinning various features, rather than storing one value for every state-action pair
