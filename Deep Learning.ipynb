#### <b><font color='blue'> Gradient Descent </font></b>
Algorithm for minimizing loss when training neural network:<br>
Start with a random choise of weights, repeat:
* calculate the gradient based on <b><font color='green'>ALL data points</font></b> (expensive part of the algorithm) direction that will lead to decreasing loss
* update weights according to the gradient

#### <b><font color='blue'> Stochastic Gradient Descent </font></b>
Algorithm for minimizing loss when training neural network:<br>
Start with a random choise of weights, repeat:
* calculate the gradient based on <b><font color='green'>ONE data point</font></b> (faster but less accurate) direction that will lead to decreasing loss
* update weights according to the gradient

#### <b><font color='blue'> Mini-Batch Gradient Descent </font></b>
Algorithm for minimizing loss when training neural network:<br>
Start with a random choise of weights, repeat:
* calculate the gradient based on <b><font color='green'>one small batch data points</font></b> (the middle ground) direction that will lead to decreasing loss
* update weights according to the gradient

#### <b><font color='blue'>Multilayer Neural Network </font></b>
Artificial neural network with an input layer, an output layer, and <b><font color='green'>at least one hidden layer</font></b>.

#### <b><font color='blue'>Backpropagation</font></b>
Algorithm for training neural networks with hidden layers. Start with a random choie of weights. Repeat:<br>
- Calculate error for output layer
- For each layer, starting with output layer, and moving inwards towards earliest hidden layer:

    - Propagate error back one layer
    - Update weights

#### <b><font color='blue'>Deep Neural Networks</font></b>
Neural network with <b><font color='green'>multiple hidden layers</font></b>.

#### <b><font color='blue'>Dropout</font></b>
Temporarily removing units -- selected at random -- from a neural network to prevent over-reliance on certain units / to prevent overfitting.

# <b> <font color=#ff4600>2. Convolutional Neural Networks (CNN) / Computer Vision</font></b><br>

Computational methods for analysing and understanding digital images. e.g. <br>
- face recognition in social media to tag friends
- autonomous driving to detect color of the traffic light
- hand writing recognition

#### <b><font color='blue'>Pooling</font></b>
Reducing the size of an input by sampling from regions in the input.

#### <b><font color='blue'>Max-Pooling</font></b>
Pooling by choosing the maximum value in each region.

#### <b><font color='green'>Convolutional Neural Network</font></b>
Neural network that has use <b><font color='green'>convolution</font></b>, usually for analysing images. (e.g. CNN mnist) <br>

<a href="geeksforgeeks.org/machine-learning/introduction-convolution-neural-network/"> Reading: Introduction to Convolution Neural Network</a>

- <b>input image</b>: some grid of pixels
- <b>convolution</b>: applying some image filters
- <b>feature maps</b>: each extacts different feature out of the input image -> adjusting value of the kernel to minimize the loss (can be repeated). e.g. 1st convolution and pooling - low-level features (e.g. edges, curves, shapes), 2nd high-levvel features (e.g. obejects, eyes)
- <b>pooling</b>: reduce the dimension of the feature map
- <b>flattening</b> -> finally traditional neural network with hidden layers

<b>What the CNN of mnist dataset does:</b>
- Minimize the loss and maximize the accuracy
- The process learns the weight (gradient descent) and features

#### <b><font color='blue'>Feed-Forward Neural Network</font></b>
Neural network that has connections <b><font color='green'>only in one direction</font></b>.<br>
input -> network -> output

# <b> <font color=#ff4600>3. Recurrent Neural Network (RNN)</font></b><br>

Neural network that generates output that feeds back into its own inputs.
Examples:
- Microsoft - Caption Bot
- Youtube dectect copyrights violations / video classification (genre: education / culture / music)
- Time Series Analysis
