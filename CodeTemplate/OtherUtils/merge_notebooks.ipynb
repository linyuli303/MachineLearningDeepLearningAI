```python
dir =r"Enter_Your_Directory"
```

```python
import os
os.chdir(dir)
```

```python
ls = []
for f in os.listdir():
    if f.endswith(".ipynb"):
        ls.append(f)

for nb in ls:
    print(f'"{nb}"')
```

```python
import nbformat

notebooks = ls
merged_nb = None

for nb_file in notebooks:
    with open(nb_file, "r", encoding="utf-8") as f:
        nb = nbformat.read(f, as_version=4)
        if merged_nb is None:
            merged_nb = nb
        else:
            # Append all cells except the first notebook's metadata
            merged_nb.cells.extend(nb.cells)

with open("merged.ipynb", "w", encoding="utf-8") as f:
    nbformat.write(merged_nb, f)

print("Notebooks merged into merged.ipynb")
```
